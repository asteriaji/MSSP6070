```python
# 2.b.
import pandas as pd

aapl = pd.read_csv("data/Week 05 Examples/aapl.csv")
amzn = pd.read_csv("data/Week 05 Examples/amzn.csv")
nflx = pd.read_csv("data/Week 05 Examples/nflx.csv")
fb = pd.read_csv("data/Week 05 Examples/fb.csv")
goog = pd.read_csv("data/Week 05 Examples/goog.csv")
```

```python
# 2.c.
aapl['ticker'] = "AAPL"
amzn['ticker'] = "AMZN"
nflx['ticker'] = "NFLX"
fb['ticker'] = "FB"
goog['ticker'] = "GOOG"

display(aapl.head())
display(amzn.head())
display(nflx.head())
display(fb.head())
display(goog.head())
```

```python
# 2.d.
df = pd.concat([aapl, amzn, nflx, fb, goog], ignore_index=True)
df
```

```python
# 2.e.
faanf = pd.concat([aapl, amzn, nflx, fb, goog], ignore_index=True)
faanf.to_csv("faanf.csv", index=False)
faanf.head()
```

```python
# 2.f.
faanf['date'] = pd.to_datetime(faanf['date'])
display(faanf['date'].dtype)
faanf['volume'] = faanf['volume'].astype(int)
display(faanf['volume'].dtype)

faanf = faanf.sort_values(by=['date', 'ticker'])
display(faanf.head())
```

```python
# 2.g.
faanf = faanf.sort_values(by=['volume'], ascending=False)
print('The 7 rows within the highest value for volume:')
display(faanf.head(7))
```
