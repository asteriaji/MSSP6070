```python
# Normality: Shapiro-Wilk test
from scipy.stats import shapiro

def check_normality(data, group_var, outcome_var):
    print(f"\nNormality: {outcome_var} by {group_var}")

    for group in data[group_var].dropna().unique():
        values = data.loc[data[group_var] == group, outcome_var].dropna()

        stat, p = shapiro(values)

        print(f"{group}: N={len(values)}, " f"W={stat:.4f}, p={p:.4g}")

check_normality(df, 'EDU_FATHER_GROUP', 'MAT_S11')
```

```python
# Normality: Q-Q plots
import scipy.stats as stats
import matplotlib.pyplot as plt

def plot_qq(data, group_var, outcome_var):
    for group in data[group_var].dropna().unique():
        values = data.loc[data[group_var] == group, outcome_var].dropna()

        stats.probplot(values, dist="norm", plot=plt)
        plt.title(f"Q-Q Plot: {outcome_var} - {group}")
        plt.show()

plot_qq(df, 'EDU_FATHER_GROUP', 'MAT_S11')
# Normality is assessed using the Shapiro-Wilk test and Q-Q plots. The
# Shapiro-Wilk test indicates statistically significant deviations from
# normality across all three father-education groups (p < .05). However,
# given the large sample sizes and the sensitivity of the Shapiro-Wilk test
# to small deviations from normality, Q-Q plots are also examined. The
# Q-Q plots show that the observations generally follow the theoretical
# normal line, particularly in the central region, although some deviations
# are visible in the tails. Overall, the results suggest that the
# normality assumption is reasonably satisfied, despite some deviations
# from normality in the tails.
```

```python
# To determine whether there is a statistically significant difference
# across the three groups, a one-way ANOVA is conducted.

from statsmodels.formula.api import ols
import statsmodels.api as sm

# CC_S11
model_cc = ols(
    'CC_S11 ~ C(EDU_FATHER_GROUP)',
    data=df
).fit()

sm.stats.anova_lm(model_cc, typ=2)
# The one-way ANOVA shows a statistically significant difference in
# citizen competencies performance across three father's education groups,
# F(2, 12408) = 104.56, p < .001.
```

```python
# Effect size is calculated to examine how much father's education influence
# outcomes.
anova_table = sm.stats.anova_lm(model_cc, typ=2)

ss_between = anova_table.loc['C(EDU_FATHER_GROUP)', 'sum_sq']
ss_total = anova_table['sum_sq'].sum()

eta_squared = ss_between / ss_total

print(f"Eta squared (η²) = {eta_squared:.4f}")
# The father's education level accounts for approximately 1.66% of the
# variance in students' citizen competencies performance.
```

```python
# Tukey's HSD post-hoc test is used to identify which pairs of groups
# differed significantly.
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(
    endog=df['CC_S11'],
    groups=df['EDU_FATHER_GROUP'],
    alpha=0.05)

print(tukey)
# Post-hoc Tukey HSD tests show that Complete and Incomplete groups
# do not differ significantly in citizen competencies performance (p = .80).
# In contrast, Complete group has significantly higher mean scores than
# Other/Unclear group (mean difference = 2.90, p < .001), and
# Incomplete group also has significantly higher mean scores than the
# Other/Unclear group (mean difference = 2.64, p < .001).
```
