### sample variance
Remember to use Bessel's correction.

$$
s^2 = \frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}$$

```python
sample_variance = sum((salary - mean_salaries)**2 for salary in salaries)/ (100 - 1)
print(sample_variance)
```

```python
# Verify results by import the statistics module
from statistics import variance
variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

$$
s = \sqrt{\frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}} = \sqrt{s^2}
$$

```python
sample_standard_deviation = variance(salaries)**0.5
print(sample_standard_deviation)
```

```python
# Verify results by import the statistics module
from statistics import stdev
stdev(salaries)
```

## Exercise 6: Calculating more statistics
### range

```python
range_salaries = max(salaries) - min(salaries)
print(range_salaries)
```

### coefficient of variation

```python
from statistics import stdev
coefficient_of_variation = stdev(salaries) / mean_salaries
print(coefficient_of_variation)
```

### interquartile range

```python
from statistics import quantiles
Q1 = quantiles(salaries, n=4, method='inclusive')[0]
Q3 = quantiles(salaries, n=4, method='inclusive')[2]
interquartile_range = Q3 - Q1
print (interquartile_range)
```

### quartile coefficent of dispersion

```python
from statistics import quantiles
quartile_coefficent_of_dispersion = (Q3 - Q1) / (Q3 + Q1)
print(quartile_coefficent_of_dispersion)
```

## Exercise 7: Scaling data
### min-max scaling

$$
x_{scaled}=\frac{x - min(X)}{range(X)}
$$

```python
scaled_salaries = [(salary - min(salaries)) / range_salaries for salary in salaries]

print(scaled_salaries)
```

### standardizing

```python
stdev_salaries = stdev(salaries)
standardizing_salaries = [(salary - mean_salaries)/stdev_salaries for salary in salaries]
print(standardizing_salaries)
```
