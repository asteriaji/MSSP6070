```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
DATA_PATH = Path("/content/MSSP6070/data/Student Performance Data.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("Student Performance Data.csv")
spds = pd.read_csv(DATA_PATH)
```

```python
spds.head()
```

```python
race_group_A = lambda data: data[data['race/ethnicity'] == 'group A']
new_spds1 = race_group_A(spds)
print(new_spds1)
```

```python
mean_score = lambda data: data[['math score', 'reading score', 'writing score']
                               ].mean(axis=1).to_frame(name='average score')
new_spds2 = mean_score(spds)
print(new_spds2)
```
