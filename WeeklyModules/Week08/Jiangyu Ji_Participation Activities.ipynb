```python
df = pd.read_csv('/content/MSSP6070/data/InternetSales.csv', delimiter= ',', parse_dates=['BirthDate','DateFirstPurchase','OrderDate','DueDate', 'ShipDate'],encoding='cp1252', low_memory=False, index_col='ID')
```

```python
date_cols = ["OrderDate", "DueDate", "ShipDate"]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")

# Check invalid date formats
for col in date_cols:
    print(f"{col} parsing failures:", df[col].isna().sum())
```

```python
internet_sales = df.assign(
    orderyear=df['OrderDate'].dt.year,
    orderquarter=df['OrderDate'].dt.quarter,
    ordermonth=df['OrderDate'].dt.month,
    orderday=df['OrderDate'].dt.day)
```

```python
quarterly_sales = (
    internet_sales
    .groupby(['orderyear', 'orderquarter'])['SalesAmount']
    .sum())

print(quarterly_sales)
# Since 2010 and 2014 only include data from one querter of the year,
# these two are excluded from later on analysis.
```

```python
internet_sales.query(
    '2010-12-31 < OrderDate',
    inplace=True)

internet_sales.query(
    'OrderDate < "2013-12-29"',
    inplace=True)
```
