```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
DATA_PATH = Path("/content/MSSP6070/data/parsed_origin.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("parsed_origin.csv")
df = pd.read_csv(DATA_PATH)
```

```python
# 2.a.
df.info()
percentile_95 = df[(df['parsed_place'] == 'Japan') & (df['magType'] == 'mb')]['mag'].astype(float).quantile(0.95)
print('The 95th percentile of earthquake magnitude in Japan for earthquakes with a magnitude type of mb is',percentile_95)
```

```python
# 2.b.
indonesia_number = df[df['parsed_place'] == 'Indonesia']
percentage = indonesia_number['tsunami'].value_counts()[1] / len(indonesia_number) * 100
print('The percentage of earthquakes in Indonesia that were coupled with tsunamis is', percentage)
```

```python
# 2.c.
Nevada = df[df['place'].str.contains('Nevada', na=False)]
Nevada.describe()
```

```python
# 2.d.
df['ring_of_fire'] = df['parsed_place'].str.contains(
    r'|'.join([
        'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
        'Guatemala', '^Mexico', 'Japan', 'Philippines', 'Indonesia',
        'New Zealand', 'Antarctic', 'Canada', 'Fiji',
        'Alaska', 'Washington', 'California', 'Russia',
        'Taiwan', 'Tonga', 'Kermadec Islands']),
    regex=True,
    na=False)
df

```

```python
# 2.e.
in_ring_of_fire_number = df[
    (df['ring_of_fire'] == True) &
    (df['type'] == 'earthquake')
]

print('The number of earthquakes in the Ring of Fire locations is', len(in_ring_of_fire_number))

not_in_ring_of_fire_number = df[
    (df['ring_of_fire'] == False) &
    (df['type'] == 'earthquake')
]

print('The number of earthquakes in the Ring of Fire locations is', len(not_in_ring_of_fire_number))
```

```python
# 2.f.
tsunami_ring_of_fire = df[(df['ring_of_fire'] == True) & (df['tsunami'] == 1)]
print(len(tsunami_ring_of_fire))
```
