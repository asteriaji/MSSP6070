### sample variance
Remember to use Bessel's correction.

$$
s^2 = \frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}$$

```python
sample_variance = sum((salary - mean_salaries)**2 for salary in salaries)/ (100 - 1)
print(sample_variance)
```

```python
from statistics import variance
variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

$$
s = \sqrt{\frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}} = \sqrt{s^2}
$$

```python
sample_standard_deviation = variance(salaries)**0.5
print(sample_standard_deviation)
```

```python
from statistics import stdev
stdev(salaries)
```

```python
from statistics import quantiles
Q1 = quantiles(salaries, n=4, method='inclusive')[0]
Q3 = quantiles(salaries, n=4, method='inclusive')[2]
interquartile_range = Q3 - Q1
print (interquartile_range)
```

```python
from statistics import quantiles
quartile_coefficent_of_dispersion = (Q3 - Q1) / (Q3 + Q1)
print(quartile_coefficent_of_dispersion)
```
