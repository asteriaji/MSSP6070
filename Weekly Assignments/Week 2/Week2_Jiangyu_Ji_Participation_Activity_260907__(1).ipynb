```python
def comma_space_and(items):
    if len(items) == 1:
        return items[0]
    return ", ".join(items[0:-1]) + ", and " + items[-1]
spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_space_and(spam))
```
